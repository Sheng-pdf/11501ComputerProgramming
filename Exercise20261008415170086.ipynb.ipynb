n = int(input("請輸入正整數 n："))

for i in range(1, n + 1):
    for j in range(1, n + 1):
        print(f"{i}x{j}={i*j:<2}", end="  ")
    print()

1. 只能使用while。                                                                                            
3. 字串格式化只能使用 f-string 對齊。                                                                       

n = int(input("請輸入正整數 n："))

i = 1
while i <= n:
    j = 1
    while j <= n:
        print(f"{i:03b}x{j:03b}={(i*j):06b}", end="  ")
        j += 1
    print()
    i += 1
